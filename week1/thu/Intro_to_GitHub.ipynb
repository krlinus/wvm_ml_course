# **Git and GitHub: An Introduction**

Before learning about GitHub, it is important to understand **Git**, the technology that GitHub is built upon.

## **Why Do We Need Version Control?**

Imagine you are working on a programming assignment and save multiple versions of your project as you make changes:

```text
Project.cpp
Project_v2.cpp
Project_final.cpp
Project_final2.cpp
Project_final_really_final.cpp
```

After a while, it becomes difficult to remember which version contains the latest changes or to recover an earlier version if something goes wrong.

A **Version Control System (VCS)** solves this problem by tracking changes to your files over time. Instead of creating many copies of the same project, a version control system keeps a complete history of your work, allowing you to:

* Track changes made to your project
* Restore previous versions if needed
* Organize your work more efficiently
* Safely back up your projects
* Collaborate with others on software projects

<br>

# **What is Git?**

**Git** is a free and open-source distributed version control system created by **Linus Torvalds** in 2005. It allows developers to keep track of changes made to their projects and maintain a complete history of their work.

Unlike traditional file storage, Git records snapshots of your project whenever you save important changes. These snapshots allow you to return to previous versions if necessary.

Git runs **locally on your computer**, meaning you can use it even without an internet connection.

Some of Git's key capabilities include:

* Tracking changes to files
* Saving project history
* Restoring previous versions
* Managing multiple versions of a project
* Supporting collaboration among developers

<br>

# **What is GitHub?**

While Git manages your project **locally**, **GitHub** is a cloud-based platform that stores Git repositories online.

GitHub allows you to:

* Back up your projects online
* Access your work from different computers
* Share your projects with others
* Showcase your programming portfolio
* Collaborate with other developers

GitHub is one of the most widely used platforms for software development and is used by individuals, universities, startups, and many of the world's largest technology companies.

<br>

# **Git vs. GitHub**

Although their names are similar, **Git** and **GitHub** are different tools that work together.

| Git                        | GitHub                                |
| -------------------------- | ------------------------------------- |
| Version control software   | Cloud platform for Git repositories   |
| Installed on your computer | Accessed through a web browser        |
| Tracks project history     | Stores and shares repositories online |
| Can be used without GitHub | Built around Git repositories         |

A simple way to remember the difference is:

> **Git manages your project. GitHub stores and shares your project.**

# **Essential Git Terminology**

*  ## **Repository (Repo)**

    A **repository**, or **repo**, is a project's folder that contains all of your source code, files, and the complete history of changes.

    Think of a repository as the home for your project.



*  ## **Commit**

    A **commit** is a saved checkpoint of your project.

    Each commit records the changes made since the previous commit and includes a short message describing what was changed.

    Example commit messages:

      * Initial project setup
      * Added menu functionality
      * Fixed input validation
      * Updated README



*  ## **Push**

    A **push** uploads your local commits from your computer to GitHub so they are stored online.



*  ## **Clone**

    A **clone** creates a local copy of an existing GitHub repository on your computer.

# **Working with GitHub**

There are two common ways to work with GitHub:

1. **Using the GitHub Website (Web Interface)** – No software installation required. This is the easiest method for beginners.
2. **Using Git from the Command Line** – The standard workflow used by professional software developers.

For this course, you may use either method unless instructed otherwise.

<br>

## **Method 1: Using the GitHub Website (Recommended for Beginners)**

#### **Step 1. Create a GitHub Account**

1. Go to https://github.com
2. Click **Sign Up**.
3. Create your account.
4. Verify your email address.
5. Log in to GitHub.

<br>

#### **Step 2. Create a New Repository**

1. Click the **+** icon in the upper-right corner.
2. Select **New repository**.
3. Enter a repository name.

Example:

```
ml-project1
```

4. Choose **Public**.
5. Check **Add a README file**.
6. Click **Create repository**.

<br>

#### **Step 3. Upload Your Project Files**

1. Open your repository.
2. Click **Add file**.
3. Select **Upload files**.
4. Drag and drop your project files.
5. Wait for the upload to finish.

<br>

#### **Step 4. Commit Your Changes**

At the bottom of the page:

Commit message:

```
Initial project upload
```

Click:

**Commit changes**

Your files are now saved in GitHub.

<br>

#### **Step 5. Update Your Project Later**

Whenever you make changes:

1. Click **Add file**
2. Upload the updated files.
3. Enter a commit message.

Example:

```
Added input validation
```

4. Click **Commit changes**.

Repeat this process throughout the semester.


<br>

This downloads the repository to your computer.

<br>

#### **Step 2. Move into the Repository**

```bash
cd cpp-project1
```

<br>

#### **Step 3. Add or Modify Files**

Copy your project files into the repository folder or edit the existing files.

<br>

#### **Step 4. Check the Status**

```bash
git status
```

Git shows which files have changed.

<br>

#### **Step 5. Stage the Files**

```bash
git add .
```

This stages all modified files.

<br>

#### **Step 6. Commit the Changes**

```bash
git commit -m "Added menu functionality"
```

A commit creates a snapshot of your work.

<br>

#### **Step 7. Upload the Changes**

```bash
git push
```

Your commits are uploaded to GitHub.

<br>

## **Typical Git Workflow**

The standard Git workflow is:

```text
Edit Files
      ↓
git status
      ↓
git add .
      ↓
git commit -m "Describe your changes"
      ↓
git push
      ↓
GitHub
```
