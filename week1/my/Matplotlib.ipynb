# **Introduction to Matplotlib**

Welcome to the world of data visualization with Matplotlib!

If you've been working with data using libraries like NumPy, you've likely encountered situations where seeing your data visually would be incredibly helpful.

That's where Matplotlib comes in.

##**What is Matplotlib?**

Matplotlib is a powerful and flexible plotting library for Python, allowing you to create a wide variety of static, interactive, and even animated visualizations directly from your Python code.

##**Why use Matplotlib?**

Matplotlib works seamlessly with NumPy arrays, making it easy to plot your data.

Whether you need to
* see trends in your data,
* compare different datasets, or
* present your findings,


Matplotlib provides the tools to do so effectively.






---



**Basic Concepts**

While Matplotlib can do a lot, the core concepts are quite straightforward. We'll dive deeper into these later, but for now, let's briefly touch upon the most fundamental ideas:

*   **Figure:** Imagine the Figure as the entire window or canvas where you draw your plot. It's the top-level container for everything you see.
*   **Axes:** Within a Figure, you have one or more Axes. An Axes is the area where the actual data is plotted. It contains the x-axis, y-axis, labels, title, and the data points or lines themselves. You can think of an Axes as an individual plot within the Figure.
*   **Plot Types:** Matplotlib supports many different types of plots, such as line plots (great for showing trends over time), scatter plots (useful for seeing the relationship between two variables), bar plots (good for comparing categories), and many more.

Don't worry if these concepts aren't completely clear yet. We'll explore them through hands-on examples in the following steps. The main takeaway is that Matplotlib provides a structured way to build your visualizations, starting with a canvas (Figure) and then adding your plots (Axes) to it.

Get ready to bring your data to life visually!



# **Matplotlib Basics**

## **Creating a Simple Line Plot**

Line plot is one of the most common types of plots and is great for visualizing trends or relationships between two variables.

We'll use NumPy to generate some sample data and Matplotlib to draw the plot.

### **Importing**

Import the `matplotlib.pyplot` module under the name `plt` (the tidy way):

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Generate x values using np.linspace
x = np.linspace(0, 15, 10)

# Generate y values using a simple function (e.g., y = x^2)
y = x**2

# Create the line plot: simply pass x and y values to this function.
plt.plot(x, y)
```

Depending on your environment, you may need to use plt.show() to display the plot

```python
# Display the plot
plt.show()
```

### **Adding label & title**

Pass the desired label & title as a string argument.

```python

plt.plot(x, y)
plt.xlabel('x values')
plt.ylabel('y values (x^2)')
plt.title('Simple Line Plot of y = x^2')
```

### **Editing more figure parameters**

```python
plt.plot(x, y)
plt.xlabel('x values')
plt.ylabel('y values (x^2)')
plt.title('Simple Line Plot of y = x^2')
plt.xlim(0,15) # Lower Limit, Upper Limit
plt.ylim(0,200) # Lower Limit, Upper Limit
```

## **Exporting a plot**

```python
plt.plot(x,y)
plt.xlabel('x values')
plt.ylabel('y values (x^2)')
plt.title('Simple Line Plot of y = x^2')
plt.savefig('example.png')
```

## **Matplotlib Styling**

### **Setting colors, linewidths, linetypes**

Matplotlib gives you *a lot* of options for customizing colors, linewidths, and linetypes.



**Line color:** You can easily change the color of your line plot using the `color` argument within the `plt.plot()` function. You can specify colors by name (e.g., 'red', 'blue', 'green') or using hexadecimal color codes.

```python
plt.plot(x, y, color='green')
plt.plot(x+2, y, color='b')
plt.plot(x+4, y, color="#8B008B")
```

**Linewidth:** To change the line width, we can use the `linewidth` or `lw` keyword argument.

```python
plt.plot(x, y, linewidth=0.25)
plt.plot(x+2, y, lw=2)
plt.plot(x+4, y, lw=5)
```

**Linestyle:** There are many linestyles to choose from, here is the selection

```python
plt.plot(x, y,  lw=2, linestyle='-')  # solid
plt.plot(x+2, y, lw=2, ls='-.')       # dash and dot
plt.plot(x+4, y, lw=2, ls=':')        # dots
plt.plot(x+6, y, lw=2, ls='--')       # dashes
```

### **Markers**

In addition to lines, you can add markers to your plot to highlight the individual data points. This can be particularly useful when you have a limited number of data points or want to emphasize their exact locations.

You can add markers using the `marker` argument within the `plt.plot()` function. Matplotlib provides a variety of marker styles, such as 'o' for circles, 'x' for crosses, '*' for stars, and many more.


```python
plt.plot(x, y, marker='+',markersize=10)
plt.plot(x+2, y, marker='o',ms=10)
plt.plot(x+4, y, marker='1',ms=10)
plt.plot(x+6, y, marker='s',ms=10,lw=0)   # make linewidth zero to see only markers
```

## **Different Plot Types**

So far, we've focused on line plots, which are excellent for showing trends. However, Matplotlib offers a rich variety of plot types, each suited for different kinds of data and insights.

Let's explore two more common and useful plot types

### **Scatter Plots**

**When to use a scatter plot:**

A scatter plot is ideal for visualizing the relationship between two numerical variables. Each point on the plot represents a single data point, with its position determined by its values for the two variables. Scatter plots can help you identify:

*   **Correlations:** Do the variables tend to increase or decrease together?
*   **Clusters:** Are there groups of data points that are similar?
*   **Outliers:** Are there data points that are far away from the others?



```python
# Generate random data for a scatter plot
x_scatter = np.random.rand(50) * 10
y_scatter = x_scatter + np.random.randn(50) * 2
```

```python
# Create the scatter plot
plt.scatter(x_scatter, y_scatter)

# Add a title and axis labels
plt.title('Scatter Plot of Random Data')
plt.xlabel('X values')
plt.ylabel('Y values')
```

**Notice:** What will happen if we use line plot on scattered data?

```python
plt.plot(x_scatter, y_scatter)
```

### **Bar Plots**

**When to use a bar plot:**

A bar plot is used to compare the values of different categories. It represents each category with a rectangular bar, where the height or length of the bar is proportional to the value it represents. Bar plots are great for:

*   **Comparing quantities:** Easily see which category has the highest or lowest value.
*   **Showing distributions:** Understand how values are distributed across different categories.
*   **Tracking changes over time for discrete categories:** While line plots are often better for continuous time series, bar plots can be used for discrete time points or periods.


To create a bar plot, you can use the `plt.bar()` function. This function typically takes two main arguments: the categories (often a list of strings or numbers representing the x-axis) and the corresponding values (representing the heights of the bars).


```python
# Generate data for a bar plot
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 12, 39]

# Create the bar plot
plt.bar(categories, values)

# Add a title and axis labels
plt.title('Bar Plot of Categories')
plt.xlabel('Category')
plt.ylabel('Value')
```

###**Matplotlib OOP API**

  * Matplotlib’s object-oriented (OOP) API treats plots as Figure objects—essentially a blank canvas.
  * Axes objects are added to the figure, and plotting is done on these axes.
  * This layered approach provides fine-grained control over every aspect of the plot.

### **Sub Plots**

The plt.subplots() object will act as a more automatic axis manager. This makes it much easier to show multiple plots side by side.

Note how we use tuple unpacking to grba both the Figure object and a numpy array of axes:


```python
import numpy as np

a = np.linspace(0,10,11)
b = a ** 4

x = np.arange(0,10)
y = 2 * x
```

```python
import matplotlib.pyplot as plt
# Use similar to plt.figure() except use tuple unpacking to grab fig and axes
fig, axes = plt.subplots()

# Now use the axes object to add stuff to plot
axes.plot(x, y, 'r')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title'); #; hides Out[]
```

### **Adding rows and columns**

Then you can specify the number of rows and columns when creating the subplots() object:

```python
# Empty canvas of 1 by 2 subplots
fig, axes = plt.subplots(nrows=1, ncols=2)
```

```python
# Axes is an array of axes to plot on
axes
```

```python
axes.shape
```

```python
# Empty canvas of 2 by 2 subplots
fig, axes = plt.subplots(nrows=2, ncols=2)
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

for axe in axes:
    axe.plot(x,y)
```

```python
fig,axes = plt.subplots(nrows=1,ncols=2)

axes[0].plot(a,b)
axes[1].plot(x,y)
```

```python
# NOTE! This returns 2 dimensional array
fig,axes = plt.subplots(nrows=2,ncols=2)

axes[0][0].plot(a,b)
axes[1][1].plot(x,y)
```

A common issue with matplolib is overlapping subplots or figures. We ca use **fig.tight_layout()** or **plt.tight_layout()** method, which automatically adjusts the positions of the axes on the figure canvas so that there is no overlapping content:

```python
# NOTE! This returns 2 dimensional array
fig,axes = plt.subplots(nrows=2,ncols=2)

axes[0][0].plot(a,b)
axes[1][1].plot(x,y)

plt.tight_layout()
```

### **Parameters on subplots()**

Recall we have both the Figure object and the axes. Meaning we can edit properties at both levels.

```python
fig,axes = plt.subplots(nrows=2,ncols=2,figsize=(12,8))

# SET YOUR AXES PARAMETERS FIRST

# Parameters at the axes level
axes[0][0].plot(a,b)
axes[0][0].set_title('0 0 Title')


axes[1][1].plot(x,y)
axes[1][1].set_title('1 1 Title')
axes[1][1].set_xlabel('1 1 X Label')

axes[0][1].plot(y,x)
axes[1][0].plot(b,a)

# THEN SET OVERALL FIGURE PARAMETERS

# Parameters at the Figure level
fig.suptitle("Figure Level",fontsize=16)


plt.show()
```

## Putting it all together

### Subtask:
Create a final example that combines several of the concepts covered in the tutorial.


```python
# This is a markdown cell introducing the final combined example.

"""
## Putting It All Together: A Combined Example

We've covered the basics of creating plots, customizing them with titles, 
labels, colors, and markers, and explored different plot types like scatter 
and bar plots. Now, let's create a final example that combines several of 
these concepts into a single figure, potentially using subplots to display 
different visualizations side-by-side.

This example will demonstrate how to create a more complex and informative 
visualization by layering different elements and plot types.
"""

# Generate new sample data for the combined example
np.random.seed(42) # for reproducibility

# Data for a line plot (e.g., a sine wave with some noise)
x_line = np.linspace(0, 10, 100)
y_line = np.sin(x_line) + np.random.randn(100) * 0.1

# Data for a scatter plot (e.g., related to the line data but with more variation)
x_scatter = np.random.rand(50) * 10
y_scatter = np.sin(x_scatter) + np.random.randn(50) * 0.5

# Data for a second scatter plot (e.g., another set of related data)
x_scatter2 = np.random.rand(50) * 10
y_scatter2 = np.cos(x_scatter2) + np.random.randn(50) * 0.3
```

```python
# Create a figure and a set of subplots
fig, axes = plt.subplots(2, 1, figsize=(10, 8)) # 2 rows, 1 column

# Plot the line data on the first subplot
axes[0].plot(x_line, y_line, color='blue', linestyle='--', label='Sine Wave with Noise')
axes[0].set_title('Line Plot: Sine Wave with Noise')
axes[0].set_xlabel('X values')
axes[0].set_ylabel('Y values')
axes[0].legend()

# Plot the first scatter data on the second subplot
axes[1].scatter(x_scatter, y_scatter, color='red', marker='o', label='Related Scatter Data 1')

# Plot the second scatter data on the second subplot
axes[1].scatter(x_scatter2, y_scatter2, color='purple', marker='x', label='Related Scatter Data 2')

axes[1].set_title('Scatter Plots: Related Data')
axes[1].set_xlabel('X values')
axes[1].set_ylabel('Y values')
axes[1].legend()

# Add a title for the overall figure
fig.suptitle('Combined Plot Example', fontsize=16)

# Adjust layout to prevent titles/labels from overlapping
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to make space for suptitle

# Display the combined plot
plt.show()
```
