## **Step 5: Apply / Lambda**
1. Create a new column total_guests = adults + children + babies using:
    - A custom function with apply()
    - A lambda function with apply()
2. Create a new column booking_type:
    - total_guests <= 2 → Small Party
    - total_guests > 2 → Large Party

```python
#'total_guests' in df.columns

df['bkg_type']=df.apply(lambda x: 'Large Party' if sum([x['adults'],x['children'],x['babies']])>2 else 'Small Party',axis=1 )
```

```python
df.columns
```

```python
def calc_total_guests(rec):
    return sum([rec['adults'],rec['children'],rec['babies']])
df['total_guests']=df.apply(calc_total_guests,axis=1)
```

```python
df.columns
```

```python
df[df['total_guests']>=7].filter(['bkg_type','total_guests'])
```

```python
import matplotlib.pyplot as pyl

ng_intervals=list(range(0,61,5))
ng_int_labels=[f'{x}-{x+4}' for x in ng_intervals]
print(f'ng_intervals={ng_intervals}')
print(f'ng_labels={ng_int_labels}')
df['numguest_group']=pd.cut(df['total_guests'],
                            bins=ng_intervals[2:],
                            labels=ng_int_labels[2:-1],
                            right=False
                            )
counts=df['numguest_group'].value_counts().sort_index()
#print(df)
print(f'counts={counts}')
f,p=pyl.subplots(1,1)
counts.plot(kind='bar', ax=p)
#p.bar(ng_intervals[:-1],counts)
```
