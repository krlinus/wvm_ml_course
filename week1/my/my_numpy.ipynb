```python
import pandas as pd
s1=pd.Series([10,20,30],index=['A','B','C'])
s2=pd.Series([1,2,3],index=['B','C','D'])
print(s1+s2)
```

```python
df=pd.DataFrame({"A":[5,8,9],"B":[1,2,3]},index=[10,20,30])
print(df.loc[20,"A"])
```

```python
df=pd.DataFrame({
    "Name":["Amy","Bob","Chris","Dean"],"Score":[90,80,90,70]
})

result = (df[df["Score"] >= 80].sort_values("Score")).drop_duplicates(subset=["Score"]))

print(result["Name"].tolist())
```

```python
df=pd.DataFrame({"Name":["A","B","C","D"],"Score":[70,95,80,95]})

df = df.sort_values("Score")
print(df)
print(df.iloc[0])
print(df.iloc[0]["Name"])
```
