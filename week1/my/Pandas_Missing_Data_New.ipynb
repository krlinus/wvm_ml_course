# **Handling Missing**

Real-world datasets often contain missing values. Many statistical methods and machine learning models **cannot handle missing data**, so we must decide how to deal with it.

In Pandas, missing values usually appear as:
- `NaN` (Not a Number) for numeric data
- `pd.NaT` for missing datetime values

There is **no single correct solution** for handling missing data. The best choice depends on the dataset, the problem, and the trade-offs you are willing to make.


## **Main Options for Handling Missing Data**

### **Keep Missing Data (Do Nothing)**

**Description:**   Leave missing values as `NaN`.

**Pros:**
- Easiest option.
- Does not change or manipulate the original data.

**Cons:**
- Many machine learning models cannot work with missing values.
- You may lose useful information if a reasonable replacement exists.

This can be acceptable for exploratory analysis or models that explicitly support missing values.

### **Drop Missing Data**

**Description:** Remove rows or columns that contain missing values.

**Pros:**

- Simple and fast.
- Pandas allows rule-based dropping (e.g., drop rows with too many missing values).

**Cons:**

- Can remove a large portion of the dataset.
- May limit future models if dropped features become available later.

<br>

### **Fill (Replace) Missing Data**

**Description:** Replace missing values with reasonable estimates.

**Pros:**

- Preserves more data.
- Often necessary for machine learning.

**Cons:**

- Requires assumptions.
- You are technically "making up" values.

<br>

**Common Strategies**

 - **Fill with a constant**
 - **Fill placeholders with meaningful values**
 - **Fill with statistical values:** Filled values could be based on averages, medians, or domain knowledge.

 - **Interpolate Missing Data:** Estimate missing values using nearby values.

  Best for:
  - Time series

  Cons:
  - Requires strong assumptions.
  - Not appropriate for unordered datasets.

### **Key Takeaways**

- Dropping missing data can remove too much information.
- Filling missing values preserves data but introduces assumptions.
- Interpolation should only be used when row order is meaningful.

There is no single correct strategy — the best choice depends on context.


# **Handling Missing Data in Pandas**

<br>

## **1. Understanding Missing Data**

Pandas represents missing values in different ways depending on the data type:

| Representation | Data Type | Notes                                                                      |
| -------------- | --------- | -------------------------------------------------------------------------- |
| `np.nan`       | float     | Standard NumPy NaN for numeric data                                        |
| `None`         | object    | Python None for object types                                               |
| `pd.NaT`       | datetime  | Missing datetime values                                                    |
| `pd.NA`        | new types | Unified missing indicator for nullable integer, boolean, and string dtypes |


```python
import numpy as np
import pandas as pd

np.nan      # float missing value
pd.NA       # nullable data type missing value
pd.NaT      # datetime missing value
```

**Important:** Avoid direct comparisons with missing values. For example:

```python
myvar = np.nan
```

```python
myvar == np.nan    # False
```

```python
myvar in [np.nan]  # True
```

```python
myvar is np.nan    # True
```

Missing values are “unknown,” so logic comparisons may not behave as expected.


## **2. Example Dataset**
Suppose we have a dataset where people scored actors before and after watching their movies:

```python
df = pd.read_csv('movie_scores.csv')
df.head()
```

```python
import pandas as pd
import numpy as np
```

## **3. Detecting Missing Values**

**Check for missing values:**


```python
df.isnull()    # True where values are missing
```

```python
df.notnull()   # True where values exist
```

**Select rows with non-missing values:**

```python
df[df['first_name'].notnull()]
```

**Select rows with missing pre_movie_score but known sex:**

```python
df[(df['pre_movie_score'].isnull()) & (df['sex'].notnull())]
```

## **4. Dropping Missing Data**
**Drop rows with any missing values:**


```python
df.dropna()
```

**Drop columns with any missing values:**

```python
df.dropna(axis=1)
```

```python
df
```

**Drop rows only if a minimum number of non-NA values is present:**

```python
df.dropna(thresh=4)  # Keep rows with at least 4 non-NA values
```

**Define in which columns to look for missing values:**

```python
df.dropna(subset=['last_name'])
```

**Tip:** Check help(df.dropna) for more options like subset, how, and inplace.

## **5. Filling Missing Data**

You can replace missing values with a fixed value or computed statistics:

**Fill with a constant value:**


```python
df.fillna("NEW VALUE!")
```

**Fill a specific column:**

```python
df['first_name'] = df['first_name'].fillna("Empty")
```

```python
df['pre_movie_score'] = df['pre_movie_score'].fillna(0)
```

**Fill numeric missing values with the mean:**

```python
df['pre_movie_score'] = df['pre_movie_score'].fillna(df['pre_movie_score'].mean())
```

**Fill all numeric columns with their respective means:**

```python
numeric_cols = df.select_dtypes(include='number').columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
```

## **6. Interpolating Missing Data**

Interpolation estimates missing values based on neighboring data. Use this method carefully—it should make sense for your dataset.

```python
airline_tix = {'first':100, 'business':np.nan, 'economy-plus':50, 'economy':30}
ser = pd.Series(airline_tix)

ser.interpolate()                # Linear interpolation
ser.interpolate(method='linear')
```

## **7. Key Takeaways**

- pd.NA is a unified missing value for multiple data types.

- Avoid direct comparison between missing values.

- Use isnull()/notnull() to detect missing data.

- Drop rows or columns with dropna() when appropriate.

- Fill missing values with fillna() for simple replacement or computed statistics.

- Interpolate only if it makes sense for your dataset.
