- body_mass_g – Body mass (grams)
- sex – Sex of the penguin

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('penguins.csv')
```

```python
df.head()
```

```python
df.groupby('species')['body_mass_g'].mean().to_frame()
```

There are 3 species of penguin in the list. The Gentoo is the biggest one

```python
df.groupby(['island','species'])['species'].value_counts().to_frame()
```

The Adelie species, which is the smallest, reside in all three islands. The Gentoo is found in Biscoe and the Chinstrap is found in Dream
The Gentoo are the highest number of penguin, but all put together the Adelie outnumber them

```python
df[['flipper_length_mm','body_mass_g']].corr()
```

There appears to be a positive correlation between the flipper length and the body mass

```python
df[['bill_length_mm','body_mass_g']].corr()
```

The correlation between the bill length and body mass is relatively lesser

```python
df['sex'].unique()
```

There appears to be some missing, invalid or unavaiable information about the genders of the animals

```python
df.groupby(['island','species','sex'])['sex'].count().to_frame()
```

There appears to be a remarkably balanced sex ratio of the penguins in all three islands

```python
df[df['sex'].isnull()].groupby(['island','species']).count()
```

There are the locations and species of the penguins with missing gender data.
