# **Bias–Variance Trade-Off: Overfitting and Underfitting**

The **bias–variance trade-off** describes the balance between a model’s ability to fit training data and its ability to generalize to unseen data. This concept applies to all supervised learning algorithms and becomes especially apparent when increasing model complexity, such as in polynomial regression.

### **Overfitting (High Variance)**
Overfitting occurs when a model is excessively complex and captures noise in the training data rather than the underlying pattern.
- Very low training error
- High test/validation error
- Poor generalization to unseen data  
This can be misleading, as strong training performance may falsely suggest a good model.

### **Underfitting (High Bias)**
Underfitting occurs when a model is too simple to capture the underlying structure of the data.
- High training error
- High test/validation error
- Poor performance on both seen and unseen data  
Underfitting is generally easier to detect than overfitting.

### **Model Complexity and Error**
As model complexity increases:
- **Training error** typically decreases monotonically.
- **Test error** decreases initially, reaches a minimum, and then increases as overfitting begins.

The optimal model complexity is achieved at the point where test error is minimized, not where training error is lowest.

### **Practical Model Selection**
When visual inspection is not possible (e.g., high-dimensional data), model evaluation relies on:
- Train/test splits or cross-validation
- Plots of error versus model complexity

In polynomial regression, complexity corresponds to polynomial degree. In other algorithms, complexity may relate to hyperparameters (e.g., number of trees in a random forest).

### **Key Takeaway**
A well-performing model balances bias and variance, achieving low and comparable errors on both training and test datasets. Selecting model complexity based on test performance is essential for reliable generalization.

### **Summary of Overfitting and Underfitting in Regression**

**Overfitting (High Variance):** The model is too complex and fits the noise in the training data, leading to excellent training performance but poor generalization on new data. Causes include:


1.  **Too many features/high dimensionality:** if you include a very large number of features (variables) in your linear regression model, especially when the number of features approaches or exceeds the number of data points, the model can start to "memorize" the training data rather than learning general patterns.

    Each additional feature adds another dimension to the model, increasing its complexity.

2.  **Irrelevant features (noise):** Model learns patterns from non-predictive data.
3.  **Insufficient data:** If your training dataset is small, a linear model  might pick up on spurious correlations that are unique to that small dataset. When exposed to new data, these spurious correlations don't hold, leading to poor performance.
4.  **High-degree polynomial features:** Creates excessive flexibility in polynomial regression.
5.  **Gradient Descent with very small learning rate & many iterations:** Model over-optimizes training error, fitting noise.
6.  **Lack of regularization (L1/L2):** Allows model (OLS/Gradient Descent) to assign large coefficients to noise.

**Underfitting (High Bias):** The model is too simple to capture the underlying patterns in the data, leading to poor performance on both training and new data. Causes include:
1.  **Too few features:** Model lacks predictive power.
2.  **Model is too simplistic:** For example, using a linear model for highly non-linear data.
3.  **Gradient Descent with very large learning rate:** Model overshoots minimum, fails to converge.
4.  **Gradient Descent with too few iterations:** Model hasn't learned enough from the data.
5.  **Too much regularization:** Over-penalizes coefficients, making the model too simple.

<br><br>

Overfitting and underfitting are not specific to regression; they can affect any ML model, including classification or deep learning models. We will discuss these scenarios as we learn each model. While the core concepts of overfitting and underfitting remain the same, their manifestations and specific causes can vary slightly across different model types.
