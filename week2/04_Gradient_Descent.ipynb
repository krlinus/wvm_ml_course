## **Simple Linear Regression - Simple Gradient Descent Implementation**

**Example (Students Scenario)**




| hours_studied| final_grade (%) |                
| ---------------------: | --------------: |
|                      3 |              68 |
|                      6 |              80 |
|                     10 |              94 |
|                      8 |              87 |

<br>

| x | y |                
| ---------------------: | --------------: |
|                      3 |              68 |
|                      6 |              80 |
|                     10 |              94 |
|                      8 |              87 |

<br>


**Resulting model:**
$$
\hat{y} = \beta_0 + \beta x
$$


```python
import numpy as np

# -----------------------------
# Dataset: hours studied vs final grade
# -----------------------------
x = np.array([3, 6, 10, 8], dtype=float)  # hours_studied
y = np.array([68, 80, 94, 87], dtype=float)  # final_grade
m = len(y)  # number of samples

# -----------------------------
# Initialize coefficients
# -----------------------------
beta_0 = 0.0  # intercept
beta_1 = 0.0  # slope

# -----------------------------
# Hyperparameters
# -----------------------------
alpha = 0.01  # learning rate
num_iterations = 6000

# -----------------------------
# Gradient Descent Loop
# -----------------------------
for i in range(num_iterations):
    # 1. Compute predictions
    y_hat = beta_0 + beta_1 * x  # element-wise multiplication

    # 2. Compute gradients using sums / dot products
    gradient_0 = (1/m) * np.sum(y_hat - y)           # derivative w.r.t beta_0
    gradient_1 = (1/m) * np.sum((y_hat - y) * x)     # derivative w.r.t beta_1

    # 3. Update coefficients
    beta_0 = beta_0 - alpha * gradient_0
    beta_1 -= alpha * gradient_1

    # Optional: print loss every 500 iterations
    if i % 500 == 0:
        mse = (1/m) * np.sum((y_hat - y)**2)
        print(f"Iteration {i:4d}: beta_0 = {beta_0:.4f}, beta_1 = {beta_1:.4f}, MSE = {mse:.4f}")

# -----------------------------
# Final coefficients
# -----------------------------
print("\nFinal model:")
print(f"Intercept (beta_0): {beta_0:.4f}")
print(f"Slope (beta_1): {beta_1:.4f}")

# -----------------------------
# Make predictions
# -----------------------------
y_pred = beta_0 + beta_1 * x
print("\nPredictions:")
for xi, yi in zip(x, y_pred):
    print(f"Hours studied: {xi}, Predicted final grade: {yi:.2f}")
```

## **Simple Linear Regression - Matrix Form Gradient Descent Implementation**

```python
import numpy as np

# --------------------------
# Step 1: Prepare the data
# --------------------------

# Feature: hours_studied
X = np.array([3, 6, 10, 8], dtype=float).reshape(-1, 1)

# Target: final_grade
y = np.array([68, 80, 94, 87], dtype=float).reshape(-1, 1)

# Add intercept term (column of ones)
X = np.hstack([np.ones((X.shape[0], 1)), X])  # shape (4,2)

# Initialize beta coefficients (β0, β1)
beta = np.zeros((X.shape[1], 1))  # shape (2,1)

# --------------------------
# Step 2: Gradient Descent
# --------------------------
learning_rate = 0.01  # can be larger for 1 feature
num_iterations = 6000

m = X.shape[0]  # number of samples

for i in range(num_iterations):
    # Prediction
    y_hat = X @ beta

    # Compute gradient: ∇J(β) = (1/m) * X^T * (y_hat - y)
    gradient = (1/m) * (X.T @ (y_hat - y))

    # Update beta coefficients
    beta -= learning_rate * gradient

    # Print loss every 100 iterations
    if i % 100 == 0:
        loss = (1/(2*m)) * np.sum((y_hat - y)**2)
        print(f"Iteration {i}: MSE Loss = {loss:.4f}")

# --------------------------
# Step 3: Print results
# --------------------------
print("\nLearned coefficients (beta):")
print(f"β0 (intercept) = {beta[0,0]:.4f}")
print(f"β1 (slope)     = {beta[1,0]:.4f}")

# Predictions
y_pred = X @ beta
print("\nPredicted final grades:")
print(y_pred)
```

## **Multiple Linear Regression - Simple Gradient Descent Implementation**

```python
import numpy as np

# -----------------------------
# Dataset: multiple features
# -----------------------------
hours_studied = np.array([3, 6, 10, 8], dtype=float)
homework_avg  = np.array([65, 78, 92, 85], dtype=float)
attendance    = np.array([70, 85, 95, 88], dtype=float)
y             = np.array([68, 80, 94, 87], dtype=float)  # final_grade
m = len(y)

# -----------------------------
# Initialize coefficients
# -----------------------------
beta_0 = 0.0  # intercept
beta_1 = 0.0  # hours_studied
beta_2 = 0.0  # homework_avg
beta_3 = 0.0  # attendance

# -----------------------------
# Hyperparameters
# -----------------------------
alpha = 0.0001  # learning rate (smaller because features have larger values)
num_iterations = 60000

# -----------------------------
# Gradient Descent Loop
# -----------------------------
for i in range(num_iterations):
    # 1. Compute predictions
    y_hat = beta_0 + beta_1*hours_studied + beta_2*homework_avg + beta_3*attendance

    # 2. Compute gradients
    gradient_0 = (1/m) * np.sum(y_hat - y)
    gradient_1 = (1/m) * np.sum((y_hat - y) * hours_studied)
    gradient_2 = (1/m) * np.sum((y_hat - y) * homework_avg)
    gradient_3 = (1/m) * np.sum((y_hat - y) * attendance)

    # 3. Update coefficients
    beta_0 -= alpha * gradient_0
    beta_1 -= alpha * gradient_1
    beta_2 -= alpha * gradient_2
    beta_3 -= alpha * gradient_3

    # Optional: print loss every 1000 iterations
    if i % 1000 == 0:
        mse = (1/m) * np.sum((y_hat - y)**2)
        print(f"Iteration {i:4d}: beta_0={beta_0:.4f}, beta_1={beta_1:.4f}, beta_2={beta_2:.4f}, beta_3={beta_3:.4f}, MSE={mse:.4f}")

# -----------------------------
# Final coefficients
# -----------------------------
print("\nFinal model coefficients:")
print(f"Intercept (beta_0): {beta_0:.4f}")
print(f"beta_1 (hours_studied): {beta_1:.4f}")
print(f"beta_2 (homework_avg): {beta_2:.4f}")
print(f"beta_3 (attendance): {beta_3:.4f}")

# -----------------------------
# Make predictions
# -----------------------------
y_pred = beta_0 + beta_1*hours_studied + beta_2*homework_avg + beta_3*attendance
print("\nPredictions:")
for h, hw, att, yi in zip(hours_studied, homework_avg, attendance, y_pred):
    print(f"Hours: {h}, Homework: {hw}, Attendance: {att}, Predicted Grade: {yi:.2f}")
```

## **Multiple Linear Regression - Matrix Form Gradient Descent Implementation**

```python
import numpy as np

# --------------------------
# Step 1: Prepare the data
# --------------------------

# Features: hours_studied, homework_avg, attendance
X = np.array([
    [3, 65, 70],
    [6, 78, 85],
    [10, 92, 95],
    [8, 85, 88]
], dtype=float)

# Target: final_grade
y = np.array([68, 80, 94, 87], dtype=float).reshape(-1, 1)

# Add intercept term (column of ones) to X
X = np.hstack([np.ones((X.shape[0], 1)), X])  # shape (4,4)

# Initialize beta coefficients (β0, β1, β2, β3)
beta = np.zeros((X.shape[1], 1))  # shape (4,1)

# --------------------------
# Step 2: Gradient Descent
# --------------------------
learning_rate = 0.0001  # small step size
num_iterations = 60000  # number of iterations

m = X.shape[0]  # number of samples

for i in range(num_iterations):
    # Prediction
    y_hat = X @ beta  # shape (4,1)

    # Compute gradient: ∇J(β) = (1/m) * X^T * (y_hat - y)
    gradient = (1/m) * (X.T @ (y_hat - y))

    # Update beta coefficients
    beta -= learning_rate * gradient

    # Optionally, print loss every 1000 iterations
    if i % 1000 == 0:
        loss = (1/(2*m)) * np.sum((y_hat - y)**2)
        print(f"Iteration {i}: MSE Loss = {loss:.4f}")

# --------------------------
# Step 3: Print results
# --------------------------
print("\nLearned coefficients (beta):")
for idx, b in enumerate(beta):
    print(f"β{idx} = {b[0]:.4f}")

# Predictions
y_pred = X @ beta
print("\nPredicted final grades:")
print(y_pred)
```
