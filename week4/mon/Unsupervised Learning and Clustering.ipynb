# **Unsupervised Learning and Clustering**
## **A Shift in Paradigm**

Up to this point, we have focused on supervised learning, where models are trained using labeled data. In contrast, unsupervised learning represents a fundamental shift:

- There are no labels (no 𝑌)
- We are not predicting outcomes
- Instead, we aim to discover hidden structure in the data

This shift makes unsupervised learning both:

- More flexible, and
- More ambiguous

If machine learning lies at the intersection of mathematics and art, unsupervised learning leans more heavily toward the artistic side, since there is no single “correct” answer.

## **Supervised vs. Unsupervised Learning**

- **Supervised Learning**
  - Uses labeled data (𝑋,𝑌)
  - Goal: learn a mapping 𝑋→𝑌
  - Tasks:
      - Regression (e.g., house prices)
      - Classification (e.g., tumor type)

- **Unsupervised Learning**
  - Uses only features 𝑋(no labels)
  - Goal: discover patterns or structure
  - Tasks:
      - Clustering → grouping similar data points
      - Dimensionality Reduction → simplifying features

## **Key Challenge: No Ground Truth**

A central challenge in unsupervised learning:

> How do we evaluate performance without true labels?

In supervised learning, we used:

  - Accuracy
  - MAE / MSE / RMSE
  - Precision / Recall

In unsupervised learning:

  - These metrics do not apply
  - There is no ground truth for comparison

Instead, evaluation becomes:

  - Interpretive
  - Context-dependent
  - Based on reasonableness and structure

## **Machine Learning Pipeline (Unsupervised Setting)**

The general pipeline remains similar:

  1. Data Collection
      - Sensors, surveys, logs, etc.
  2. Data Storage
      - Databases, CSV, cloud storage
  3. Data Cleaning & EDA
  4. Modeling

However, the key difference:

  - The dataset contains **only features 𝑋**

Thus:
  - No prediction task
  - Instead → **pattern discovery**

## **Two Main Branches of Unsupervised Learning**
- Clustering
  - Goal: group similar data points
  - Output: clusters (pseudo-labels)
  - Example:
      - Customer segmentation
      - Document grouping

- Dimensionality Reduction
  - Goal: reduce number of features
  - Combine features into **new components**
  - Example:
      - Reducing 25 features → 2 components for visualization
