# **Principal Component Analysis (PCA)**


In this section, we shift from **clustering** to another major class of unsupervised learning techniques:

> **Dimensionality Reduction**

Unlike clustering, which attempts to discover labels, dimensionality reduction focuses on:
- Simplifying data
- Improving interpretability
- Preserving important structure

## **Motivation for PCA**

Consider a dataset with **30+ features**.

Challenges:
- Difficult to visualize beyond 3 dimensions
- Hard to identify important features
- Complex relationships between variables

### **Key Goals of Dimensionality Reduction:**
1. **Reduce the number of dimensions**
2. **Preserve as much information (variance) as possible**
3. **Improve human understanding and visualization**


## **Important Clarification**

PCA does **NOT**:
- Select a subset of original features

PCA **DOES**:
- Create **new features (components)**
- Each component is a **combination of original features**

## **What is PCA? (High-Level Idea)**

> PCA transforms data into a new coordinate system where:
- The first axis captures the **maximum variance**
- The second axis captures the **next highest variance**
- And so on...

These new axes are called:
- **Principal Components**

## **Intuition with 2D Example**

Suppose we have two features:
- X₁
- X₂

Instead of using these axes, PCA finds a new axis:
- A **diagonal direction** capturing the most spread

### **Key Idea:**
- Project data onto this new axis
- Reduce from 2D → 1D while preserving most information

## **Variance Explained**

- **Variance = spread of data**
- PCA prioritizes directions with **maximum variance**

### **Trade-off:**
- More components → more information retained
- Fewer components → simpler representation

Example:
- 1 component may explain 90% of variance
- Remaining components explain the rest

##  **Mathematical Formulation**

Each principal component is a **linear combination** of original features:

$$Z₁ = a₁X₁ + a₂X₂ + ... + aₚXₚ$$

Where:
- Z₁ = first principal component
- a₁, a₂, ... = coefficients

##  **PCA Workflow**

- ### **Step 1: Standardize Data**
  - Center data (mean = 0)
  - Important for fair comparison


- ### **Step 2: Compute Covariance Matrix**
  - Measures relationships between features


- ### **Step 3: Compute Eigenvalues & Eigenvectors**
  - **Eigenvectors → directions (principal components)**
  - **Eigenvalues → importance (variance explained)**


- ### **Step 4: Sort Components**
  - Rank by eigenvalues (highest first)


- ### **Step 5: Select Top Components**
  - Choose number of dimensions (e.g., 2 or 3)


- ### **Step 6: Project Data**
  - Transform original data onto selected components


##  **Geometric Interpretation**

- PCA = **rotation of coordinate system**
- Aligns axes with directions of maximum variance
- Then optionally **drops lower-variance dimensions**

## **Key Concepts**

- ### **Eigenvectors**
  - Directions that remain unchanged under transformation
  - Represent principal components

- ### **Eigenvalues**
  - Measure importance (variance explained)

## **Key Takeaways**

- PCA is an **unsupervised transformation technique**
- It creates **new features**, not subsets
- It prioritizes **variance**
- It enables **efficient and interpretable representations**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cancer_tumor_data_features.csv')
df.head()
```

## **Manual Construction of PCA**


### **Scaling Data**

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_X = scaler.fit_transform(df)
scaled_X
```

```python
# Grab Covariance Matrix
covariance_matrix = np.cov(scaled_X, rowvar=False) # rowvar=False bc we care about relationships between columns not rows
```

```python
# Get Eigen Vectors and Eigen Values
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
```

```python
# Choose som number of components
num_components=2
```

```python
# Get index sorting key based on Eigen Values
sorted_key = np.argsort(eigen_values)[::-1][:num_components]
```

```python
# Get num_components of Eigen Values and Eigen Vectors
eigen_values, eigen_vectors = eigen_values[sorted_key], eigen_vectors[:, sorted_key]
```

```python
# Dot product of original data and eigen_vectors are the principal component values
# This is the "projection" step of the original points on to the Principal Component
principal_components=np.dot(scaled_X,eigen_vectors)
```

```python
principal_components
```

```python
plt.figure(figsize=(8,6))
plt.scatter(principal_components[:,0],principal_components[:,1])
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```

```python
from sklearn.datasets import load_breast_cancer
cancer_dictionary = load_breast_cancer()
cancer_dictionary.keys()
```

```python
cancer_dictionary['target']
```

```python
plt.figure(figsize=(8,6))
plt.scatter(principal_components[:,0],principal_components[:,1],c=cancer_dictionary['target'])
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```

## **PCA with Scikit-Learn**


### **Scaling Data**

```python
df = pd.read_csv('cancer_tumor_data_features.csv')
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_X = scaler.fit_transform(df)
scaled_X
```

```python
from sklearn.decomposition import PCA
help(PCA)
```

```python
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_X)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(principal_components[:,0],principal_components[:,1])
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```

```python
from sklearn.datasets import load_breast_cancer
cancer_dictionary = load_breast_cancer()
cancer_dictionary.keys()
```

```python
cancer_dictionary['target']
```

```python
plt.figure(figsize=(8,6))
plt.scatter(principal_components[:,0],principal_components[:,1],c=cancer_dictionary['target'])
plt.xlabel('First principal component')
plt.ylabel('Second Principal Component')
```

## **Fitted Model Attributes**

```python
pca.n_components
```

```python
pca.components_
```

In this numpy matrix array, each row represents a principal component, Principal axes in feature space, representing the directions of maximum variance in the data. The components are sorted by explained_variance_.

We can visualize this relationship with a heatmap:

```python
df_comp = pd.DataFrame(pca.components_,index=['PC1','PC2'],columns=df.columns)
df_comp
```

```python
plt.figure(figsize=(20,3),dpi=150)
sns.heatmap(df_comp,annot=True)
```

```python
pca.explained_variance_ratio_
```

First PCA explains 44% of the variance in my data and the second principle component explains around 19%.
Together they explain 63%.

```python
np.sum(pca.explained_variance_ratio_)
```

```python
pca_30 = PCA(n_components=30)
pca_30.fit(scaled_X)
```

```python
pca_30.explained_variance_ratio_
```

```python
np.sum(pca_30.explained_variance_ratio_)
```

```python
explained_variance = []

for n in range(1,30):
    pca = PCA(n_components=n)
    pca.fit(scaled_X)

    explained_variance.append(np.sum(pca.explained_variance_ratio_))
```

```python
plt.plot(range(1,30),explained_variance)
plt.xlabel("Number of Components")
plt.ylabel("Variance Explained");
```
