## **Hierarchical Clustering**

Hierarchical clustering aims to build a hierarchy of clusters. There are two main types:

1.  **Agglomerative (bottom-up)**: Each data point starts as a single cluster, and pairs of clusters are merged as one moves up the hierarchy.
2.  **Divisive (top-down)**: All data points start in one cluster, and the cluster is split recursively as one moves down the hierarchy.

We'll focus on agglomerative clustering, which is more common.
