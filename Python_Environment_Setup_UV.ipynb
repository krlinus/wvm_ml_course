# **Python Environment Setup for This Course (Ubuntu Mini-PCs)**

In this course, you will use Python together with several scientific libraries (NumPy, Pandas, SciPy, scikit-learn, etc.).  
To make sure everyone gets the same results and to avoid breaking the computer, we will use a **Python environment**.


### **What Is a Python Environment?**

A **Python environment** is a self-contained workspace that includes:
- A specific Python version
- A specific set of libraries
- Exact library versions

This allows:
- Everyone to run the same code
- Fewer installation problems
- No interference with the operating system

Think of it as a **sandbox** for this course.

#### **2. Local Python with Conda (Recommended for This Course)**

Conda is a tool that helps you install and manage Python **environments**. It allows you to create isolated workspaces where each project can have its own Python version and its own set of libraries, without affecting the rest of the computer.

We use Conda in this course because:

  - It works well on Linux (Ubuntu)
  - It installs scientific libraries reliably
  - It prevents conflicts between different Python versions
  - It keeps the operating system safe

With Conda, you do **not** modify the system Python that comes with Ubuntu. Instead, Conda creates a separate environment just for this course. This environment contains:

  - The correct Python version
  - All required libraries
  - Exact library versions that match the textbook

This makes sure that everyone in the class runs the same code and gets the same results.

Using the terminal may feel unfamiliar at first, but it is a standard and powerful tool in programming and data science.

### **Opening an IPython (Jupyter Notebook) File**

An **IPython file** (also called a **Jupyter Notebook**) has the file extension `.ipynb`. This type of file contains Python code, text explanations, and outputs (such as results and plots) all in one place.

To open an `.ipynb` file, you do **not** double-click it like a regular file. Instead, you open it using **JupyterLab**.

**Steps:**

1. Make sure your course environment is activated:
   ```bash
   conda activate machine-learning-book
   ```
2. Start JupyterLab:
1. Make sure your course environment is activated:
   ```bash
    jupyter lab
   ```
3. A browser window will open automatically.

4. In JupyterLab, use the file browser on the left to navigate to your course folder (for example, ml-book).

5. Click on the .ipynb file to open it.

The notebook will open in a new tab inside JupyterLab. You can run code cells by pressing Shift + Enter.

## **Python Environment Setup Using uv (macOS & Linux)**

**What Is uv?**

uv is a fast, modern Python package and environment manager written in Rust.

With uv, we can:

- Install Python (no system Python changes)
- Create a virtual environment
- Install exact package versions
- Run JupyterLab inside the environment

#### **Step 1: Install uv**

**macOS & Linux (recommended)**

Open a terminal and run:



```
curl -LsSf https://astral.sh/uv/install.sh | sh
```



Restart your terminal after installation.

Verify installation:



```
uv --version
```



### **Step 2: Create a Course Folder**

Choose where you want your course files and create a folder:



```
mkdir machine-learning-book
cd machine-learning-book
```


This folder will contain:

- Your virtual environment
- Course notebooks
- Dependency files



### **Step 3: Create a Python Virtual Environment**

Create a virtual environment named .venv:


```
uv venv
```

Activate it:

**macOS / Linux**


```
source .venv/bin/activate
```

You should now see something like:



```
(.venv) $
```







### **Step 4: Install Course Dependencies**

**Option A: Using a [requirements.txt](https://github.com/rasbt/machine-learning-book/blob/main/environment.yml) file (recommended)**

Download the requirements.txt file:



```
uv pip install -r requirements.txt
```


**Option B: Manual installation (if no file is provided)**



```
uv pip install numpy pandas scipy scikit-learn matplotlib seaborn jupyterlab
```




### **Step 5: Start JupyterLab**

Make sure your environment is activated, then run:



```
jupyter lab
```

A browser window will open automatically.



### **Step 6: Open and Run Notebooks**

- In JupyterLab, use the file browser on the left
- Navigate to your course folder (e.g., machine-learning-book)
- Click on a .ipynb file to open it
- Run cells using:
  - Shift + Enter

### **Important Notes for Students**

Always activate the environment before running Jupyter:



```
source .venv/bin/activate
```
- Do not install packages with pip install outside the environment
- If something breaks, you can delete .venv and recreate it safely
