# **Classification Performance Metrics**




After building a classification model, we need a way to **measure how good the predictions are**.

A model may produce predictions, but we must evaluate:

- How often the model is correct
- What types of mistakes it makes
- Whether those mistakes matter for the problem

To do this, we use **classification performance metrics**, which are based on something called the **confusion matrix**.

---

## **Confusion Matrix**

In a binary classification problem, there are **four possible outcomes** when comparing predictions to the true labels.

Example: Detecting whether a person has a virus.

- **1 = Infected**
- **0 = Healthy**

|                     | Actual Positive | Actual Negative |
|---------------------|----------------|----------------|
| **Predicted Positive** | True Positive (TP) | False Positive (FP) |
| **Predicted Negative** | False Negative (FN) | True Negative (TN) |

#### **Meaning of Each Case**

**True Positive (TP):**  The person is infected and the model correctly predicts infected.

**True Negative (TN):**  The person is healthy and the model correctly predicts healthy.

**False Positive (FP):**  The person is healthy but the model predicts infected.

**False Negative (FN):**  The person is infected but the model predicts healthy.

These four values form the basis for **all classification metrics**.

## **Accuracy**

Accuracy answers the question:

**How often is the model correct?**

The formula is:

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

Example:

- TP = 4
- TN = 93
- FP = 2
- FN = 1

Then:

$$
Accuracy = \frac{4 + 93}{100} = 0.97
$$

So the model is **97% accurate**.

Is this a good value for accuracy?

---

### **Accuracy Paradox**

Accuracy can be misleading when the dataset is **imbalanced**.

Example:

- 5 infected people
- 95 healthy people

If a model **always predicts healthy**, then:

- TP = 0
- TN = 95

Accuracy becomes:

$$
Accuracy = \frac{95}{100} = 95\%
$$

Even though the model **never detects infections**.

This problem is known as the **accuracy paradox**.

Because of this, we also use **precision and recall**.

---

## **Recall (Sensitivity, True Positive Rate)**

Focuses on the model's ability to correctly identify **all relevant positive cases** within a dataset.

Recall answers the question:

**When the case is actually positive, how often does the model detect it?**
or **How many relevant cases are found?**

Formula:

$$
Recall = \frac{TP}{TP + FN}
$$



Essentially, the denominator `(TP + FN)` represents **all actual positive cases** in your dataset. Recall measures the proportion of these actual positive cases that the model successfully captured.

<br>




Example:

- TP = 4
- FN = 1

$$
Recall = \frac{4}{5} = 0.8
$$

Meaning the model detects **80% of infected cases**.






#### **When & Why is Recall Important?**


Recall is important when **missing a positive case is dangerous**  
(e.g., disease detection).

Recall becomes the primary metric when the **cost of a False Negative (FN) is very high**. A False Negative means the model failed to detect a positive case that was actually present. Missing a positive case can have severe consequences, making it imperative for the model to have high recall.

#### **Example Scenario: Medical Diagnosis (Cancer Detection)**

Let's consider a model designed to detect cancer from medical scans.

*   **Positive Class:** Cancer is present.
*   **Negative Class:** Cancer is not present.

Imagine the following results for 100 patients:

*   **TP = 9:** 9 patients actually had cancer, and the model correctly identified them.
*   **FN = 1:** 1 patient actually had cancer, but the model incorrectly said they were healthy (missed the cancer).
*   **FP = 5:** 5 patients were healthy, but the model incorrectly said they had cancer.
*   **TN = 85:** 85 patients were healthy, and the model correctly identified them.

Using the Recall formula:

$$Recall = \frac{TP}{TP + FN} = \frac{9}{9 + 1} = \frac{9}{10} = 0.9$$

In this case, the model has a Recall of 90%. This means that out of all the patients who actually had cancer, the model successfully detected 90% of them. The 10% it missed (the 1 False Negative) is a serious concern, as missing a cancer diagnosis can have life-threatening implications.

#### **When to Prioritize High Recall:**

High recall is crucial in situations where:

1.  **Medical Diagnosis:** As in the cancer example, missing a disease (FN) is far more critical than a false alarm (FP).
2.  **Fraud Detection:** Failing to detect a fraudulent transaction (FN) could lead to significant financial loss for a bank or individual.
3.  **Security Systems:** Missing an intruder or a security threat (FN) can have catastrophic consequences.
4.  **Anomaly Detection:** In manufacturing, failing to identify a defective product (FN) can lead to product recalls or customer dissatisfaction.
5.  **Terrorism Screening:** Missing a potential threat (FN) at an airport is highly undesirable.

In all these scenarios, the goal is to cast a wide net and capture as many actual positive instances as possible, even if it means accepting a few more False Positives (where the model raises an alarm that turns out to be incorrect). The priority is to minimize False Negatives.

## **Precision**

Focuses on the **quality of the positive predictions made by the model**. It tells us, out of all the instances the model *predicted* as positive, how many were actually correct.

Precision answers the question:

**When the model predicts positive, how often is it correct?**

Formula:

$$
Precision = \frac{TP}{TP + FP}
$$


Essentially, the denominator `(TP + FP)` represents **all cases that the model predicted as positive**. Precision measures the proportion of these predicted positive cases that were genuinely positive.

<br>


Example:

- TP = 4
- FP = 2

$$
Precision = \frac{4}{6} = 0.67
$$

Meaning **67% of predicted infections are actually correct**.

Precision is important when **false alarms are costly**.


#### **Why is Precision Important?**

Precision becomes the primary metric when the **cost of a False Positive (FP) is very high**. A False Positive means the model incorrectly raised an alarm or identified something as positive when it was not. Such errors can lead to wasted resources, inconvenience, or even significant negative consequences.

#### **Example Scenario: Spam Email Detection**

Let's consider a model designed to detect spam emails.

*   **Positive Class:** The email is spam.
*   **Negative Class:** The email is legitimate (ham).

Imagine the following results for 100 emails:

*   **TP = 20:** 20 emails were spam, and the model correctly identified them as spam.
*   **FP = 5:** 5 emails were legitimate (ham), but the model incorrectly classified them as spam.
*   **FN = 2:** 2 emails were spam, but the model incorrectly classified them as legitimate.
*   **TN = 73:** 73 emails were legitimate, and the model correctly identified them as legitimate.

Using the Precision formula:

$$Precision = \frac{TP}{TP + FP} = \frac{20}{20 + 5} = \frac{20}{25} = 0.8$$

In this case, the model has a Precision of 80%. This means that out of all the emails the model flagged as spam, 80% were actually spam. The 20% that were incorrectly marked as spam (the 5 False Positives) are problematic because they represent important emails that a user might miss. Users would quickly lose trust in a spam filter that frequently mislabels legitimate emails as spam.

#### **When to Prioritize High Precision:**

High precision is crucial in situations where:

1.  **Spam Filtering:** As in the example, incorrectly marking a legitimate email as spam (FP) can lead to missed important communications and user frustration.
2.  **Product Recommendation Systems:** Recommending a product a customer dislikes (FP) can lead to a poor user experience and distrust in the recommendations.
3.  **Legal Document Review:** Flagging a non-relevant document as relevant (FP) means legal teams waste time reviewing unnecessary files.
4.  **Customer Service Chatbots:** If a chatbot incorrectly identifies a user's intent (FP) and provides the wrong solution, it can lead to customer dissatisfaction.
5.  **Autonomous Driving (False Obstacle Detection):** A self-driving car incorrectly identifying a non-existent obstacle (FP) could cause unnecessary braking or swerving, leading to discomfort or potential accidents.

In these scenarios, the goal is to be highly confident in every positive prediction, even if it means being more conservative and potentially missing some actual positive cases (accepting some False Negatives). The priority is to minimize False Positives.

## **Precision vs Recall Tradeoff**

There is usually a **tradeoff between precision and recall**.

Example:

- Increasing recall may increase **false positives**
- Increasing precision may increase **false negatives**

The best balance depends on the **application**.

Example:

| Problem | Important Metric |
|------|------|
Disease detection | High recall |
Spam filtering | High precision |


<br>

Imagine a model that outputs a probability score for each prediction, say, how likely an email is spam (0 to 1). To classify an email as 'spam' (positive), we use a threshold. If the probability is above the threshold, it's classified as spam; otherwise, it's not.

- **To increase Recall:** To catch more actual positive cases (spam emails), we need to be more 'lenient' with our predictions. This means lowering the prediction threshold. For example, instead of classifying emails with a spam probability > 0.8 as spam, we might lower the threshold to > 0.5.

- **The consequence of lowering the threshold:** When you lower the threshold, you're essentially expanding the net to capture more positive cases. While this helps in catching more true positives (increasing recall), it inevitably also sweeps in more cases that are actually negative but had a moderately high probability score. These are your false positives.

## **F1 Score**

The **F1-score** combines precision and recall into one metric.

It is the **harmonic mean** of precision and recall.

$$
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

The harmonic mean is used because it **penalizes extreme values**.

If either precision or recall is **very low**, the F1-score will also be low.

This gives a **balanced measure of model performance**.

## **ROC Curve**

A classifier often outputs **probabilities** rather than final labels.

To convert probabilities into classes, we choose a **threshold**.

Example:

- If probability > 0.5 → positive
- Otherwise → negative

Changing the threshold changes:

- True Positive Rate
- False Positive Rate

A **ROC (Receiver Operating Characteristic) curve** plots:

| Axis | Meaning |
|-----|-----|
X-axis | False Positive Rate |
Y-axis | True Positive Rate |

Each point on the curve represents a **different threshold**.

---

## **Area Under the Curve (AUC)**

The **Area Under the ROC Curve (AUC)** summarizes the ROC curve in one number.

- **AUC = 1** → perfect classifier
- **AUC = 0.5** → random guessing

Higher AUC means the model **better separates the classes**.

This allows us to **compare models easily**.

#### **Interpreting the Curve:**

**Trade-off:** There's usually a trade-off between precision and recall. As you increase the threshold to get higher precision (fewer false positives), you might lower the recall (miss more true positives), and vice-versa.

**Ideal Curve:** An ideal classifier would have a curve that stays as close to the top-right corner as possible (Precision = 1, Recall = 1).


**Use Cases:**

- High Precision is Critical: If the cost of false positives is very high (e.g., flagging a legitimate transaction as fraudulent), you'd want a model with high precision, even if it means sacrificing some recall.

- High Recall is Critical: If the cost of false negatives is very high (e.g., missing a cancer diagnosis), you'd want a model with high recall, even if it means accepting more false positives.

---

## **Summary**

Key classification metrics include:

| Metric | What it Measures |
|------|------|
Accuracy | Overall correctness |
Precision | Correct positive predictions |
Recall | Ability to find positive cases |
F1 Score | Balance between precision and recall |
ROC / AUC | Performance across thresholds |

These metrics help us understand **how well a classification model performs**, especially when dealing with **imbalanced datasets**.
