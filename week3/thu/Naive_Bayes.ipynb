# **Bayes' Theorem**

### **Definition**

Bayes’ Theorem allows us to compute a **conditional probability**:

$$P(A | B) = \frac {P(B | A) . P(A)} {P(B)}$$



## **What Each Term Means**

- $P(A)$: Probability of event A occuring
    
    **Prior**     → How likely event A is *before seeing any evidence*

- $P(B)$: Probability of event B occuring

    **Evidence**  → How likely event B is overall

- $P(B | A)$: Probability of event B given that A is true

    **Likelihood**  → How likely we are to observe B if A is true

- $P(A | B)$: Probability of event A given that B is true

    **Posterior**  → What we actually want: updated probability of A after observing B


## **Key Idea**

> We are **reversing the direction of probability**

We usually know: $P(B | A)$

But we want: $P(A | B)$

## **Intuition**

Bayes' theorem answers:

> “Given that I observed something (B), how should I update my belief about A?”



## **Example 1: Fire Alarm (Smoke vs Fire)**

### **Step 1: Define Events**

Let:
- A = "There is a dangerous fire"
- B = "Smoke alarm goes off"

### **Step 2: What Are We Trying to Find?**

We want: $P(Fire | Smoke)$

> **Translation:** *“If the alarm goes off, what is the probability there is actually a fire?”* or *"Probability of fire given smoke alarm"*


### **Step 3: Given Information**

- P(Fire) = 0.01  → Fires are rare (1% chance)

- P(Smoke) = 0.10  → Alarm goes off 10% of the time (including false alarms)

- P(Smoke | Fire) = 0.95  → If there is a fire, alarm detects it 95% of the time


### **Step 4: Plug Into Formula**

$P(Fire | Smoke) = \frac {P(Smoke | Fire) . P(Fire)}{P(Smoke)}$

= (0.95 * 0.01) / 0.10 = 0.0095 / 0.10 = 0.095 = 9.5%**


### **Step 5: Interpretation (Very Important)**

Even though:
- Alarm is very accurate (95%)

The probability of a real fire is only **9.5%**


### **Why Is It So Low?**

Because:
- Fires are **very rare** (low prior)
- Alarms go off **often without fire** (false positives)

> The prior probability strongly affects the result






# **From Bayes' Theorem to Naive Bayes**

### **Goal:** We want to model the probability of belonging to a given class given a vector of features:
$$P(Class | Features)$$


### **General Form:**

$$P(C_k | X) = \frac {P(C_k) P(X | C_k) }{P(X)}$$

Where:
- $C_k$ = class (e.g., positive/negative)
- $X$ = feature vector (words)


Since $P(X)$ is the same for all classes, the numerator is equivalent to a joint probability model:

$$P(C_k | X) = \frac { p(C_k, x_1,...,x_n) }{P(X)}$$

The chain rule can rewirte the numerator as a series of products of conditional probabilities:

$p(C_k, x_1,...,x_n) = p(x_1,...,x_n, C_k)$

 $ = p(x_1|x_2,...,x_n, C_k)p(x_2,...,x_n, C_k)$

 $ = p(x_1|x_2,...,x_n, C_k)p(x_2|x_3,...,x_n, C_k)p(x_3,...,x_n, C_k)$

 $= ... $

 $ = p(x_1|x_2,...,x_n, C_k)p(x_2|x_3,...,x_n, C_k) ... p(x_{n-1}|x_n, C_k) p(x_n|C_k)p(C_k)$

<br>

## **The Naive Assumption**

We assume:

Features are **mutually independent** given the class

So:

$$P(X | C_k) = P(x1 | C_k) P(x2 | C_k)  ...  P(xn | C_k)$$

$$P(C_k | X) ∝ P(C_k) \prod_{i=1}^{n} P(x_i | C_k)$$


## **Why "Naive"?**

Because in reality:
- Words are NOT independent
- Context matters

But surprisingly:
- This assumption works well in practice


## **Interpretation**

To classify a document:

1. Compute score for each class:
   $$score_k = P(C_k) ∏ P(x_i | C_k)$$

2. Choose the class with the highest score

# **Feature Extraction from Text Data**

**Why Do We Need Feature Extraction?**

Machine learning models **cannot process raw text**.

Example:
- "This movie is great" ❌ (not usable)
- [0, 1, 3, 0, ...] ✅ (numeric representation)

> Goal: Convert text → numbers


**What Are We Actually Doing?**

We are transforming: Text (documents) → Feature vectors (numbers)

So that we can compute: P(Class | Features)

<br>

### **Two Main Methods**
1. **Count Vectorization**
2. **TF-IDF (Term Frequency - Inverse Document Frequency)**
<br>



## **Part 1: Count Vectorization**

**Idea:** Represent each document by **word counts**

- **Step 1: Example Documents**

  - Document 1: "you are good"  
  - Document 2: "I feel good"  
  - Document 3: "I am good"

- **Step 2: Build Vocabulary**

  - Collect all unique words: Vocabulary = [you, are, good, I, feel, am]
  - This defines our **feature space**
  - Each word = one feature

- **Step 3: Convert to Vectors**

  - We count how many times each word appears in each document.

| Document        | you | are | good | I | feel | am |
|----------------|-----|-----|------|---|------|----|
| "you are good" |  1  |  1  |  1   | 0 |  0   | 0  |
| "I feel good"  |  0  |  0  |  1   | 1 |  1   | 0  |
| "I am good"    |  0  |  0  |  1   | 1 |  0   | 1  |

<br>

**What Does This Mean?**

- Each row = a document  
- Each column = a word (feature)  
- Each value = count of that word

**Interpretation**

We assume:

- Words are features
- Frequency = importance

**Important Observation**

  - Most values are **zero**.
  - Why? Because Not all documents contain all words
  - This creates a **sparse matrix** to save space

<br>

#### **Problems with Count Vectorization**

- **Problem 1:** Stop Words

  - Words like: "the", "a", "is" , Appear everywhere → not useful
  - **Solution:** Stop Words Removal
  - Remove very common words that: Do not carry meaningful information

- **Problem 2:** Common Words Across Documents

  - Example: Sports articles → "ball", "run"
  - These appear everywhere → not helpful for classification
  - Count Vectorization only looks at words inside ONE document
  - It ignores: How common the word is across ALL documents


<br>

## **Part 2: TF-IDF**

**Motivation**

- We want to answer: “How important is a word in this document relative to all documents?”

$$TF-IDF = TF × IDF$$

- **Term Frequency (TF):** How often a word appears in a document
  - Can be viewed as tf(t,d), calculates the frequency of each term (t) in document (d)
  - Measures **local importance** (within document)
  - TF is essentially the same idea as Count Vectorization
- **Inverse Document Frequency (IDF):** How rare a word is across all documents
  - Penalize common words (weighted less)
  - Reward rare words (weighted more)
  - Formula: $IDF = log( Total Documents / Documents containing the word )$
    - It is the logarithmically scaled ratio of the total number of documents to the number of documents containing the term.
    - If word appears in ALL documents: → IDF ≈ 0 (not important)
    - If word appears in FEW documents: → IDF is large (important)


<br>

| Word      | Appears in Docs | IDF Value | Importance |
|----------|----------------|----------|------------|
| "the"    | all            | low      | low        |
| "good"   | many           | medium   | medium     |
| "goalpost"| few           | high     | high       |

<br>

**Interpretation:** A word is important if:

- It appears frequently in the document  AND  
- It is rare across documents  



**Example Intuition:**

- Word: "goalpost" -> High TF-IDF score

  - Rare across documents → high IDF
  - Appears in one document → meaningful

- Word: "the" -> Low TF-IDF score

  - Appears everywhere → low IDF

<br>


#### **What Are We Achieving?**

We move from:

Raw text → meaningful numeric features

That:
- capture importance
- reduce noise
- improve classification

<br>

## **Building a vocabulary (Creating a "Bag of Words")**

Let's create dictionaries that correspond to unique mappings of the words in the documents. We can begin to think of this as mapping out all the possible words available for all (both) documents.

```python
with open('One.txt') as f:
    words_one = f.read().lower().split()

words_one
```

```python
len(words_one)
```

```python
uni_words_one = set(words_one)
uni_words_one
```

**Repeat for Two.txt**

```python
with open('Two.txt') as f:
    words_two = f.read().lower().split()
    uni_words_two = set(words_two)
```

```python
uni_words_two
```

**Get all unique words across all documents**

```python
all_uni_words = set()
all_uni_words.update(uni_words_one)
all_uni_words.update(uni_words_two)
all_uni_words
```

```python
full_vocab = dict()
i = 0

for word in all_uni_words:
    full_vocab[word] = i
    i = i+1
```

```python
# Do not expect this to be in alphabetical order!
# The for loop goes through the set() in the most efficient way possible, not in alphabetical order!
full_vocab
```

## **Bag of Words to Frequency Counts**

Now that we've encapsulated our "entire language" in a dictionary, let's perform *feature extraction* on each of our original documents:

**Empty counts per doc**

```python
# Create an empty vector with space for each word in the vocabulary:
one_freq = [0]*len(full_vocab)
two_freq = [0]*len(full_vocab)
all_words = ['']*len(full_vocab)
```

```python
for word in full_vocab:
    word_ind = full_vocab[word]
    all_words[word_ind] = word
```

```python
all_words
```

**Add in counts per word per doc:**

```python
# map the frequencies of each word in 1.txt to our vector:
with open('One.txt') as f:
    one_text = f.read().lower().split()

for word in one_text:
    word_ind = full_vocab[word]
    one_freq[word_ind]+=1
```

```python
one_freq
```

```python
# Do the same for the second document:
with open('Two.txt') as f:
    two_text = f.read().lower().split()

for word in two_text:
    word_ind = full_vocab[word]
    two_freq[word_ind]+=1
```

```python
two_freq
```

```python
import pandas as pd
pd.DataFrame(data=[one_freq,two_freq],columns=all_words)
```

By comparing the vectors we see that some words are common to both, some appear only in `One.txt`, others only in `Two.txt`. Extending this logic to tens of thousands of documents, we would see the vocabulary dictionary grow to hundreds of thousands of words. Vectors would contain mostly zero values, making them **sparse matrices**.


# Concepts to Consider:

## **Bag of Words and Tf-idf**
In the above examples, each vector can be considered a *bag of words*. By itself these may not be helpful until we consider *term frequencies*, or how often individual words appear in documents. A simple way to calculate term frequencies is to divide the number of occurrences of a word by the total number of words in the document. In this way, the number of times a word appears in large documents can be compared to that of smaller documents.

However, it may be hard to differentiate documents based on term frequency if a word shows up in a majority of documents. To handle this we also consider *inverse document frequency*, which is the total number of documents divided by the number of documents that contain the word. In practice we convert this value to a logarithmic scale, as described [here](https://en.wikipedia.org/wiki/Tf%E2%80%93idf#Inverse_document_frequency).

Together these terms become [**tf-idf**](https://en.wikipedia.org/wiki/Tf%E2%80%93idf).

## **Stop Words**

- Words like `"the"`, `"and"`, `"is"` appear **all the time** in almost every document.  
- They **don’t tell us anything useful** about the content.  
- So we **ignore or remove them** from our word counts.  
- This **reduces noise** and keeps the vocabulary smaller.  

<br>

## **Word Stems (Stemming)**

- Words often appear in multiple forms: `"cat"` vs `"cats"`, `"run"` vs `"running"`.  
- **Stemming** reduces words to their **root form** (`cat`, `run`).  
- This way, we **treat different forms as the same word**, shrinking the vocabulary and improving consistency.  

## **Tokenization and Tagging**

When we created our vectors the first thing we did was split the incoming text on whitespace with `.split()`. This was a crude form of *tokenization* - that is, dividing a document into individual words. In this simple example we didn't worry about punctuation or different parts of speech. In the real world we rely on some fairly sophisticated *morphology* to parse text appropriately.

Once the text is divided, we can go back and *tag* our tokens with information about parts of speech, grammatical dependencies, etc. This adds more dimensions to our data and enables a deeper understanding of the context of specific documents. For this reason, vectors become ***high dimensional sparse matrices***.

### **TfidfTransformer**

TfidfVectorizer is used on sentences, while TfidfTransformer is used on an existing count matrix, such as one returned by CountVectorizer

```python
tfidf_transformer = TfidfTransformer()
```

```python
cv = CountVectorizer()
```

```python
counts = cv.fit_transform(text)
counts
```

```python
tfidf = tfidf_transformer.fit_transform(counts)
```

```python
tfidf.todense()
```

```python
from sklearn.pipeline import Pipeline

pipe = Pipeline([('cv',CountVectorizer()),('tfidf',TfidfTransformer())])
results = pipe.fit_transform(text)
```

```python
results
```

```python
results.todense()
```
